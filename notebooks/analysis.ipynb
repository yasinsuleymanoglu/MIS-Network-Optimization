# Cloud Data Center Network Optimization Analysis

This Jupyter Notebook demonstrates the step-by-step shortest path optimization for data center routing.

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# 1. Load Data
df = pd.read_csv('../data/network_data.csv')
df.head()
```

```python
# 2. Build Graph
G = nx.Graph()
for index, row in df.iterrows():
    G.add_edge(row['source'], row['target'], weight=row['latency_ms'])

print(f"Nodes: {G.number_of_nodes()}, Edges: {G.number_of_edges()}")
```

```python
# 3. Find Shortest Path
start_node = 'DC-NewYork'
end_node = 'DC-Tokyo'
optimal_path = nx.shortest_path(G, source=start_node, target=end_node, weight='weight')
latency = nx.shortest_path_length(G, source=start_node, target=end_node, weight='weight')

print(f"Optimal Path: {' -> '.join(optimal_path)}")
print(f"Total Latency: {latency} ms")
```

```python
# 4. Visualize
plt.figure(figsize=(10, 6))
pos = nx.spring_layout(G, seed=42)
nx.draw_networkx_nodes(G, pos, node_color='lightblue', node_size=2000)
nx.draw_networkx_labels(G, pos, font_size=9, font_weight='bold')
nx.draw_networkx_edges(G, pos, edge_color='gray', width=1.5)
nx.draw_networkx_edge_labels(G, pos, edge_labels=nx.get_edge_attributes(G, 'weight'))

path_edges = list(zip(optimal_path, optimal_path[1:]))
nx.draw_networkx_nodes(G, pos, nodelist=optimal_path, node_color='lightgreen', node_size=2000)
nx.draw_networkx_edges(G, pos, edgelist=path_edges, edge_color='red', width=3.0)

plt.title("Network Optimization (Shortest Path)")
plt.axis('off')
plt.show()
```
